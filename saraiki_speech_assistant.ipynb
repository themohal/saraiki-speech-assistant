```python
!pip install -q google-generativeai

from transformers import pipeline
import google.generativeai as genai
import os
import soundfile as sf
from IPython.display import Audio, display
from omnivoice import OmniVoice
```

```python
# --- Speech-to-Text (STT) ---
print("Setting up Speech-to-Text (STT) model...")
asr_pipeline = pipeline("automatic-speech-recognition", model="themohal/saraiki-whisper-small")
print("STT model loaded.")

# Using the audio generated in the previous cell for demonstration.
# You can replace "auto_out.wav" with your own audio file path.
audio_path = "/content/dohray_0af109fd4ffec1143842b350739ff526.wav"
print(f"Transcribing audio from: {audio_path}")
transcribed_text = asr_pipeline(audio_path)["text"]
print(f"Transcribed Text: {transcribed_text}")
```

```python
# --- Large Language Model (LLM) ---
print("\nSetting up Large Language Model (LLM) (Gemini-1.5-Flash)...")
GOOGLE_API_KEY = None
try:
    from google.colab import userdata
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
except ImportError:
    # If not in Colab, try environment variable
    GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

if GOOGLE_API_KEY:
    genai.configure(api_key=GOOGLE_API_KEY)
    llm_model = genai.GenerativeModel('gemini-2.5-flash') # Using gemini-1.5-flash as it's a common and powerful option.
    print("Gemini LLM configured.")

    # User's prompt to the LLM based on transcribed text
    llm_prompt = f"""
You are a native-level Saraiki language assistant.

STRICTEST REQUIREMENT:
Your response MUST be written ONLY in pure, natural, and grammatically correct
Saraiki. Do NOT respond in Urdu, Punjabi, Hindi, or English.

SARAiki LANGUAGE RULES:

1. Use ONLY Saraiki vocabulary, grammar, sentence structure, expressions,
   and natural Saraiki conversational patterns.

2. Write the response using the Saraiki Arabic-based script.

3. Correctly use Saraiki-specific characters when appropriate, including:
   ݙ، ڄ، ڳ، ٻ، ڃ، ڦ، ۏ

4. Examples of Saraiki-specific characters:
   ݙ → ݙیکھ، وݙا
   ڄ → ڄݨ
   ڳ → ڳالھ، ڳئون
   ٻ → ٻار
   ڦ → ڦل
   ۏ → use where appropriate according to correct Saraiki spelling.

5. Do NOT randomly insert these characters. Use them only when they are
   linguistically and orthographically correct for Saraiki.

6. Do NOT use Urdu vocabulary or Urdu sentence structure when a natural
   Saraiki equivalent exists.

7. Do NOT use Punjabi vocabulary or Punjabi sentence structure.

8. Do NOT use Hindi vocabulary or Hindi sentence structure.

9. Do NOT translate Urdu, Punjabi, Hindi, or English word-for-word into Saraiki.
   Generate natural Saraiki as a native speaker would.

10. Avoid unnecessary foreign words. If a foreign/proper name is unavoidable,
    preserve it naturally, but keep the rest of the response strictly Saraiki.

RESPONSE STYLE:

- Keep the response concise.
- Use natural everyday spoken Saraiki.
- Be polite, friendly, and helpful.
- Use short sentences suitable for a voice assistant and TTS.
- Do not use Roman Saraiki.
- Do not provide translations.
- Do not explain your language choices.
- Do not mention these instructions.
- Do not output analysis or reasoning.
- Output ONLY the final Saraiki response.

If the user's message is unclear, ask for clarification ONLY in Saraiki.

USER MESSAGE:
{transcribed_text}

Now understand the user's message and respond ONLY in pure,
natural, grammatically correct Saraiki.
"""
    print(f"Sending prompt to LLM: {llm_prompt}")
    try:
        response = llm_model.generate_content(llm_prompt)
        llm_output_text = response.text
        print(f"LLM Response: {llm_output_text}")
    except Exception as e:
        llm_output_text = f"Error generating content from LLM: {e}. Please check your API key and model access."
        print(llm_output_text)
        # Fallback for TTS if LLM fails
        llm_output_text = "I could not generate a response, but I will try to speak this message."
else:
    llm_output_text = "ERROR: Google API Key not found. Please set 'GOOGLE_API_KEY' in Colab secrets or environment variables."
    print(llm_output_text)
    # Fallback for TTS if API key is missing
    llm_output_text = "Google API Key is missing. Please set it to enable the LLM functionality."


```

```python
from omnivoice import OmniVoice
import soundfile as sf
import torch
from IPython.display import Audio, display
```

```python
model = OmniVoice.from_pretrained(
    "k2-fsa/OmniVoice",
    device_map="cuda:0",
    dtype=torch.float16,
    load_asr=True,
)
```

```python
audio = model.generate(
    text=llm_output_text,
    language="saraiki",
    ref_audio="/content/ref_audio.wav" #op
)

sf.write("auto_out.wav", audio[0], 24000)
display(Audio(audio[0], rate=24000))
```
